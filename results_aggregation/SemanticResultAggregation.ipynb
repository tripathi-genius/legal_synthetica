```python
qwen3b_ft = pd.read_csv('qwen3b_finetuned_task_level_results.csv')
qwen3b_ft['model'] = 'qwen-3b-ft'
qwen3b_ft.head()
```

```python
llama3b_ft = pd.read_csv('llama3b_finetunedtask_level_results.csv')
llama3b_ft['model'] = 'llama-3b-ft'
llama3b_ft.head()
```

```python
llama3b_base = pd.read_csv('llama_3b_task_level_results.csv')
llama3b_base['model'] = 'llama-3b-base'
llama3b_base.head()
```

```python
llama8b_base = pd.read_csv('llama8btask_level_results.csv')
llama8b_base['model'] = 'llama-8b-base'
llama8b_base.head()
```

```python
llama8b_ft = pd.read_csv('llama8b_finetuned_task_level_results.csv')
llama8b_ft['model'] = 'llama-8b-ft'
llama8b_ft.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# 1. Combine your dataframes using the specified list
df_list = [qwen3b_base, qwen3b_ft, llama3b_base, llama3b_ft, llama8b_base, llama8b_ft]
combined_df = pd.concat(df_list, ignore_index=True)

# 2. Specify the tasks
tasks = [
    'argument_evaluation', 
    'arguments_generation', 
    'final_judgment', 
    'full_chain_multi_turn', 
    'full_chain_single_turn'
]

# 3. Create output directory and configure style
os.makedirs('semantic_metrics_graphs', exist_ok=True)
sns.set_theme(style="whitegrid")

metrics_to_plot = ['avg_semantic_similarity', 'avg_lexical_jaccard', 'sentence_f1_micro']
metric_labels = {
    'avg_semantic_similarity': 'Semantic Similarity',
    'avg_lexical_jaccard': 'Lexical Jaccard',
    'sentence_f1_micro': 'Sentence F1 Micro'
}

# 4. Generate multi-metric bar charts for each task
for task_name in tasks:
    task_data = combined_df[combined_df['task'] == task_name]
    
    if task_data.empty:
        continue
        
    fig, axes = plt.subplots(1, len(metrics_to_plot), figsize=(16, 5))
    fig.suptitle(f'Semantic Evaluation Metrics — {task_name.replace("_", " ").title()}', fontsize=14, fontweight='bold', y=1.03)
    
    for idx, metric in enumerate(metrics_to_plot):
        ax = axes[idx]
        sns.barplot(
            data=task_data,
            x='model',
            y=metric,
            palette='Blues_d',
            ax=ax,
            edgecolor='black'
        )
        ax.set_title(metric_labels[metric], fontsize=12, fontweight='bold')
        ax.set_xlabel('Model Variant', fontsize=10)
        ax.set_ylabel(metric_labels[metric], fontsize=10)
        ax.tick_params(axis='x', rotation=30)
        ax.set_ylim(0, 1.0)
        
    plt.tight_layout()
    filename = f'semantic_metrics_graphs/{task_name}_metrics_chart.png'
    plt.savefig(filename, dpi=300, bbox_inches='tight')
    plt.close()

print("All task-level semantic evaluation charts successfully saved!")
```
