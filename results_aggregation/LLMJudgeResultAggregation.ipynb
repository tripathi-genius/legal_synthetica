```python
import pandas as pd

qwen3b_base = pd.read_csv('qwen3b_task_level_results.csv')
qwen3b_base['model'] = 'qwen-3b-base'
qwen3b_base.head()
```

```python
qwen3b_ft = pd.read_csv('qwen3b_finetuned_task_level_results.csv')
qwen3b_ft['model'] = 'qwen-3b-ft'
qwen3b_ft.head()
```

```python
llama3b_ft = pd.read_csv('llama3b_finetuned_task_level_results.csv')
llama3b_ft['model'] = 'llama-3b-ft'
llama3b_ft.head()
```

```python
llama3b_base = pd.read_csv('llama3b_task_level_results.csv')
llama3b_base['model'] = 'llama-3b-base'
llama3b_base.head()
```

```python
llama8b_base = pd.read_csv('llama8b_task_level_results.csv')
llama8b_base['model'] = 'llama-8b-base'
llama8b_base.head()
```

```python
llama8b_ft = pd.read_csv('llama8b_finetuned_task_level_results.csv')
llama8b_ft['model'] = 'llama-8b-ft'
llama8b_ft.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# 1. Combine your dataframes
df_list = [qwen3b_base, qwen3b_ft, llama3b_base, llama3b_ft, llama8b_base, llama8b_ft]
combined_df = pd.concat(df_list, ignore_index=True)

# 2. Specify the tasks
tasks = [
    'argument_evaluation', 
    'arguments_generation', 
    'final_judgment', 
    'full_chain_multi_turn', 
    'full_chain_single_turn'
]

# 3. Create output directory and configure style
os.makedirs('evaluation_graphs_score_only', exist_ok=True)
sns.set_theme(style="whitegrid")

# 4. Generate score-only bar charts for each task
for task_name in tasks:
    task_data = combined_df[combined_df['task'] == task_name]
    
    if task_data.empty:
        continue
        
    fig, ax = plt.subplots(figsize=(8, 5))
    sns.barplot(
        data=task_data,
        x='model',
        y='mean_score_1_to_10',
        palette='Blues_d',
        ax=ax,
        edgecolor='black'
    )
    
    ax.set_title(f'LLM Judge Mean Score — {task_name.replace("_", " ").title()}', fontsize=14, fontweight='bold', pad=15)
    ax.set_xlabel('Model Variant', fontsize=11)
    ax.set_ylabel('Mean Score (1 to 10)', fontsize=11)
    ax.tick_params(axis='x', rotation=30)
    ax.set_ylim(0, 10.0)
    
    plt.tight_layout()
    filename = f'evaluation_graphs_score_only/{task_name}_score_chart.png'
    plt.savefig(filename, dpi=300, bbox_inches='tight')
    plt.close()

print("All score-only charts successfully saved!")
```
